### Split CSV File into N Chunks

This script defines a function `split_csv` that takes an input CSV file path, an output directory, and the desired number of output files (N) as arguments. It reads the CSV, calculates the chunk size, and then iterates to save each chunk as a new CSV file.

```python
import pandas as pd
import os
import math

def split_csv(input_csv_path, output_directory, num_files):
    """
    Splits a CSV file into N approximately equal smaller CSV files.

    Args:
        input_csv_path (str): The path to the input CSV file.
        output_directory (str): The directory where the output CSV files will be saved.
        num_files (int): The number of smaller CSV files to create.
    """
    if not os.path.exists(output_directory):
        os.makedirs(output_directory)

    try:
        df = pd.read_csv(input_csv_path)
        total_rows = len(df)
        # Calculate the number of rows per file, rounding up to ensure all rows are included
        rows_per_file = math.ceil(total_rows / num_files)

        for i in range(num_files):
            start_row = i * rows_per_file
            end_row = min((i + 1) * rows_per_file, total_rows)

            if start_row >= total_rows:
                break

            chunk = df.iloc[start_row:end_row]
            output_file_name = f"split_file_{i+1}.csv"
            output_file_path = os.path.join(output_directory, output_file_name)
            chunk.to_csv(output_file_path, index=False)
            print(f"Created {output_file_path} with {len(chunk)} rows.")

    except FileNotFoundError:
        print(f"Error: The file '{input_csv_path}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- Example Usage --- (You can modify these variables)
# Create a dummy CSV for demonstration if it doesn't exist
dummy_data = {'col1': range(1, 101), 'col2': [f'value_{i}' for i in range(1, 101)]}
dummy_df = pd.DataFrame(dummy_data)
dummy_csv_path = 'dummy_data.csv'
dummy_df.to_csv(dummy_csv_path, index=False)
print(f"Created dummy CSV: {dummy_csv_path}")

input_csv = dummy_csv_path  # Replace with your actual CSV file path
output_dir = 'output_csv_files' # Directory to save the split files
num_output_files = 5          # Number of files to split into

split_csv(input_csv, output_dir, num_output_files)
```
